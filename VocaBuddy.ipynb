## 📘 VocaBuddy 單字本

由 **Developer A：陳柏東（@Dong-Chen-1031）** 新增第一批英文單字與中文解釋。

每一筆單字包含：英文單字、詞性、中文解釋、例句。


```python
# VocaBuddy 單字本
# 每一筆單字包含：英文單字、中文解釋、詞性、例句

vocabulary = [
    {
        "word": "collaboration",
        "meaning": "合作、協作",
        "pos": "n.",
        "example": "Our team finished the project through close collaboration.",
    },
    {
        "word": "repository",
        "meaning": "倉庫、儲存庫",
        "pos": "n.",
        "example": "We pushed the notebook to our GitHub repository.",
    },
    {
        "word": "commit",
        "meaning": "提交（版本紀錄）、承諾",
        "pos": "v. / n.",
        "example": "Please write a clear message before you commit your changes.",
    },
    {
        "word": "notebook",
        "meaning": "筆記本、（Colab）筆記本檔案",
        "pos": "n.",
        "example": "Open the notebook in Google Colab and run all cells.",
    },
    {
        "word": "review",
        "meaning": "檢查、複習、審查",
        "pos": "v. / n.",
        "example": "The reviewer will review the README before we submit it.",
    },
]

print(f"單字本建立完成，目前共有 {len(vocabulary)} 個單字。")
```

```python
def show_vocabulary(vocab_list):
    """依序列出所有單字、詞性、中文解釋與例句。"""
    print("=" * 46)
    print("📘 VocaBuddy 單字清單")
    print("=" * 46)
    for index, item in enumerate(vocab_list, start=1):
        print(f"{index}. {item['word']} ({item['pos']})")
        print(f"   中文解釋：{item['meaning']}")
        print(f"   例句：{item['example']}")
        print("-" * 46)
    print(f"總共 {len(vocab_list)} 個單字")


show_vocabulary(vocabulary)
```

## 🎲 隨機抽取單字 + 單字測驗

由 **Developer B：@Friedturtleee** 新增的功能：

- `draw_random_word()`：隨機抽取一個單字並顯示中文解釋
- `quiz()`：以中文解釋出題的四選一測驗，並記錄答對與答錯、計算正確率、列出需要複習的單字

> 在 Colab 中想自己作答，請把最後一行改成 `quiz(vocabulary, questions=3)`（不帶 `answers`），即可手動輸入 A/B/C/D。


```python
import random


def draw_random_word(vocab_list):
    """隨機抽取一個單字並顯示中文解釋。"""
    item = random.choice(vocab_list)
    print(f"🎲 隨機單字：{item['word']} ({item['pos']})")
    print(f"   中文解釋：{item['meaning']}")
    return item


def quiz(vocab_list, questions=3, answers=None):
    """簡單的單字測驗：看中文解釋，選出正確的英文單字，並記錄答對與答錯。"""
    picked = random.sample(vocab_list, k=min(questions, len(vocab_list)))
    correct, wrong = 0, []

    for number, item in enumerate(picked, start=1):
        others = [v["word"] for v in vocab_list if v["word"] != item["word"]]
        options = random.sample(others, k=min(3, len(others))) + [item["word"]]
        random.shuffle(options)

        print(f"\n第 {number} 題：「{item['meaning']}」的英文是？")
        for label, option in zip("ABCD", options):
            print(f"   {label}. {option}")

        if answers is None:
            reply = input("請輸入選項（A/B/C/D）：").strip().upper()
        else:  # 自動作答模式，方便「全部執行」時不必手動輸入
            reply = answers[number - 1]
            print(f"請輸入選項（A/B/C/D）：{reply}")

        chosen = options["ABCD".index(reply)] if reply in "ABCD" and reply else ""
        if chosen == item["word"]:
            correct += 1
            print("   ✅ 答對了！")
        else:
            wrong.append(item)
            print(f"   ❌ 答錯了，正確答案是 {item['word']}。")

    print("\n" + "=" * 46)
    print("📊 測驗結果")
    print(f"   題數：{len(picked)}　答對：{correct}　答錯：{len(picked) - correct}")
    print(f"   正確率：{correct / len(picked) * 100:.0f}%")
    if wrong:
        print("   需要複習的單字：" + "、".join(v["word"] for v in wrong))
    print("=" * 46)
    return correct, wrong


random.seed(12)  # 固定亂數，讓「全部執行」的結果可重現
draw_random_word(vocabulary)
quiz(vocabulary, questions=3, answers=["D", "C", "A"])
```
