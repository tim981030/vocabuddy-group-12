## 📘 VocaBuddy 單字本

由 **Developer A：陳柏東（@Dong-Chen-1031）** 新增第一批英文單字與中文解釋。

每一筆單字包含：英文單字、詞性、中文解釋、例句。


```python
# VocaBuddy 單字本
# 每一筆單字包含：英文單字、中文解釋、詞性、例句

vocabulary = [
    {
        "word": "collaboration",
        "meaning": "合作、協作",
        "pos": "n.",
        "example": "Our team finished the project through close collaboration.",
    },
    {
        "word": "repository",
        "meaning": "倉庫、儲存庫",
        "pos": "n.",
        "example": "We pushed the notebook to our GitHub repository.",
    },
    {
        "word": "commit",
        "meaning": "提交（版本紀錄）、承諾",
        "pos": "v. / n.",
        "example": "Please write a clear message before you commit your changes.",
    },
    {
        "word": "notebook",
        "meaning": "筆記本、（Colab）筆記本檔案",
        "pos": "n.",
        "example": "Open the notebook in Google Colab and run all cells.",
    },
    {
        "word": "review",
        "meaning": "檢查、複習、審查",
        "pos": "v. / n.",
        "example": "The reviewer will review the README before we submit it.",
    },
]

print(f"單字本建立完成，目前共有 {len(vocabulary)} 個單字。")
```

```python
def show_vocabulary(vocab_list):
    """依序列出所有單字、詞性、中文解釋與例句。"""
    print("=" * 46)
    print("📘 VocaBuddy 單字清單")
    print("=" * 46)
    for index, item in enumerate(vocab_list, start=1):
        print(f"{index}. {item['word']} ({item['pos']})")
        print(f"   中文解釋：{item['meaning']}")
        print(f"   例句：{item['example']}")
        print("-" * 46)
    print(f"總共 {len(vocab_list)} 個單字")


show_vocabulary(vocabulary)
```
